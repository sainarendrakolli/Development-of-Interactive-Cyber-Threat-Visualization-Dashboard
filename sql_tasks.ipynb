```python
df = pd.read_csv("medical_dataset_20_records.csv")
df
```

```python
conn = sqlite3.connect("medical.db")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS medical_data (
    Age INTEGER,
    Gender TEXT,
    Blood_Pressure INTEGER,
    Cholesterol INTEGER,
    Disease TEXT
)
""")

df.to_sql("medical_data", conn, if_exists="replace", index=False)
```

```python
import pandas as pd

pd.read_sql("SELECT * FROM medical_data", conn)
```

```python
cursor.execute("SELECT * FROM medical_data")
rows = cursor.fetchall()
rows
```

```python
pd.read_sql("SELECT COUNT(*) FROM medical_data", conn)
```

```python
pd.read_sql("SELECT * FROM medical_data WHERE Age > 50", conn)
```

```python
pd.read_sql("""
SELECT Disease, COUNT(*)
FROM medical_data
GROUP BY Disease
""", conn)
```
